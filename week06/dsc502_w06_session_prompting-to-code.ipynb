### Exercise 3 — Three flawed functions

**Function A** — Read it. What's wrong?

````python
# Function A — provided as-is; do not modify until you've identified the flaw

def get_order_summary(df):
    """Returns a summary of orders by restaurant."""
    result = (
        df
        .copy()
        .assign(order_count=1)
        .groupby('restaurant_name')
        .agg(
            total_orders=('order_count', 'sum'),
            total_revenue=('order_total', 'sum'),
            avg_rating=('rating', 'mean'),
            unique_customers=('customer_name', lambda x: len(x.unique()))
        )
        .reset_index()
        .sort_values('total_revenue', ascending=False)
    )
    return result

# Run it
get_order_summary(df)
````

**Function C** — Read it. What's wrong?

````python
# Function C — provided as-is; do not modify until you've identified the flaw

def combine_order_batches(batch1, batch2):
    """Combines two batches of QuickBite orders into one DataFrame."""
    return batch1.append(batch2, ignore_index=True)

# Split the DataFrame into two halves and try to combine them
half = len(df) // 2
batch_a = df.iloc[:half]
batch_b = df.iloc[half:]

result = combine_order_batches(batch_a, batch_b)
print(f'Combined: {len(result)} rows')
````

````python
# Now run this — what happens?
import warnings
warnings.simplefilter('always')

# Try this alternative approach that pandas recommends instead
result = pd.concat([batch_a, batch_b], ignore_index=True)
print(f'Combined using pd.concat: {len(result)} rows')
````

**What's the flaw in Function C?**

*Hint: `.append()` was deprecated in pandas 1.4 and removed entirely in pandas 2.0. This Codespace runs pandas 2.x, so calling `.append()` will throw an `AttributeError` — the method no longer exists. This is a real and common problem with AI-generated pandas code: AI was trained on older code where `.append()` was valid. The correct replacement is `pd.concat([df1, df2], ignore_index=True)`.*
